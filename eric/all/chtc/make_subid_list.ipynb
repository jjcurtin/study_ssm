```python
import pandas as pd
```

```python
subject_info = pd.read_csv('/Users/eric/repos/aud/data/subject_info.csv')
subject_info.head()
```

```python
export_df = subject_info.query("responsiveness>=0.25").copy(deep=True).reset_index(drop=True)
print(export_df)
```

```python
#fit_types = ['MAP_mle','MAP_marss','MLE','MARSS']
#fit_types = ['MAP_mle','MLE']
fit_types = ['MAP_mle']
f = open('subid_list.txt','w')
f.close()
f = open('subid_list.txt','a')
for i,row in export_df.iterrows():
    id = int(row.subid)
    for fit in fit_types:
        if(i==export_df.index[-1] and fit==fit_types[-1]):
            f.write("{} {}".format(id,fit))
        else:
            f.write("{} {}\n".format(id,fit))
f.close()
```
