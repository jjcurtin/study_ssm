```python
import numpy as np
import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
demo_path = "/Volumes/private/studydata/risk/data_processed/shared/screen.csv"

subject_info = pd.read_csv('/Users/eric/repos/aud/data/subject_info.csv')
included_subids = subject_info.query("responsiveness>=0.25").subid.unique()

raw_demo_df = pd.read_csv(demo_path)
demo_df = raw_demo_df.query("subid in @included_subids")[['dem_1','dem_2','dem_3','dem_4','dem_7','auh_5',
                       'dsm5_1', 'dsm5_2', 'dsm5_3', 'dsm5_4', 'dsm5_5', 
                       'dsm5_6', 'dsm5_7', 'dsm5_8', 'dsm5_9', 'dsm5_10', 'dsm5_11']]
dsm_cols = ['dsm5_1', 'dsm5_2', 'dsm5_3', 'dsm5_4', 'dsm5_5', 
                       'dsm5_6', 'dsm5_7', 'dsm5_8', 'dsm5_9', 'dsm5_10', 'dsm5_11']
print(raw_demo_df.shape)
print(demo_df.shape)
```

```python
def yn(val):
    if val=='Yes':
        return 1
    elif val=='No':
        return 0
    else:
        return "err"
```

```python
demo_df.rename(columns={'dem_1':'age','dem_2':'sex','dem_3':'race','dem_4':'hlso','dem_7':'income','auh_5':'quit_attempts'},inplace=True)

demo_df['dsm_count']=demo_df.apply(lambda row: yn(row.dsm5_1)+yn(row.dsm5_2) +yn(row.dsm5_3)+yn(row.dsm5_4)
                                   +yn(row.dsm5_5)+yn(row.dsm5_6)+yn(row.dsm5_7)+yn(row.dsm5_8)
                                   +yn(row.dsm5_9)+yn(row.dsm5_10)+yn(row.dsm5_11),axis=1)
```

```python
tot = demo_df.shape[0]

# Age
print("Age, m: {:.3f}, sd: {:.3f}".format(demo_df.age.mean(),demo_df.age.std()))

# Sex 
sex_counts = demo_df.sex.value_counts()
print("Sex, Male: {} {:.3f}%, Female: {} {:.3f}".format(sex_counts.Male,sex_counts.Male/(tot),
                                                        sex_counts.Female,sex_counts.Female/(tot)))

# Race
race_counts = demo_df.race.value_counts()
print("AIAN: {} {:.3f}, A: {} {:.3f}, BAA: {} {:.3f}, WC: {} {:.3f}, OM: {} {:.3f}".format(
    race_counts['American Indian/Alaska Native'],race_counts['American Indian/Alaska Native']/tot,
    race_counts['Asian'],race_counts['Asian']/tot,
    race_counts['Black/African American'],race_counts['Black/African American']/tot,
    race_counts['White/Caucasian'],race_counts['White/Caucasian']/tot,
    race_counts['Other/Multiracial'],race_counts['Other/Multiracial']/tot
))

# HLS origin
hlso_counts = demo_df.hlso.value_counts()
print("HLSO: Yes {} {:.3f}, No {} {:.3f}".format(
    tot-hlso_counts['No, I am not of Hispanic, Latino, or Spanish origin'],(tot-hlso_counts['No, I am not of Hispanic, Latino, or Spanish origin'])/tot,
    hlso_counts['No, I am not of Hispanic, Latino, or Spanish origin'],hlso_counts['No, I am not of Hispanic, Latino, or Spanish origin']/tot,
))

# Personal income
print("Income, m: {:.3f}, sd: {:.3f}".format(demo_df.income.mean(),demo_df.income.std()))

# Quit attempts
print("Quit attempts, m: {:.3f}, sd: {:.3f}".format(demo_df.quit_attempts.mean(),demo_df.quit_attempts.std()))

# Symptom count
print("Symptom count, m: {:.3f}, sd: {:.3f}".format(demo_df.dsm_count.mean(),demo_df.dsm_count.std()))
```
